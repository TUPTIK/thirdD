```python
import numpy as np
import matplotlib.pyplot as plt
import json


style = dict(
    marker="o",
    ms=4,
    linestyle="solid",
    linewidth=2,
    color="black"
)
```

```python
class ThirdD:
    def __init__(self, poligons: list, rotations: list):
        self.p = poligons
        self.rot = rotations

    def poligons(self):
        for a in range(len(self.p)-1):
            self.pol(a)

    def pol (self, a:int):
        pol = self.p[a]
        x = []
        y = []
        for p in pol:
            self.x = 0
            self.y = 0
            self.xA(p[0])
            self.yA(p[1])
            self.zA(p[2])
            x.append(self.x)
            y.append(self.y)
        plt.plot(x,y,**style)
        plt.fill_between(x,y,color="orange")

    def xA(self,a):
        t = 3**0.5
        self.x += a*t/2 
        self.y -= a*0.5

    def yA(self,b):
        self.y += b

    def zA(self,c):
        t = 3**0.5
        self.x -= c*t/2
        self.y -= c*0.5
```

```python
a = 15
b = 20
c = 10
pol = [
    [[0,0,0],
    [a,0,0],
    [a,b,0],
    [0,b,0],
    [0,0,0]],
    [[0,0,0],
    [0,b,0],
    [0,b,c],
    [0,0,c],
    [0,0,0]],
    [[0,0,0],
    [0,0,c],
    [a,0,c],
    [a,0,0],
    [0,0,0]],
    [[a,b,c],
    [a,0,c],
    [a,0,0],
    [a,b,0],
    [a,b,c]],
    [[a,b,c],
    [0,b,c],
    [0,b,0],
    [a,b,0],
    [a,b,c]],
    [[a,b,c],
    [a,0,c],
    [0,0,c],
    [0,b,c],
    [a,b,c]]
]

with open("3d.txt","w") as file:
    file.write(json.dumps(pol))
```

```python
with open("3d.txt") as file:
    pol = json.loads(file.read())
```

```python
p = ThirdD(pol,[0,0,0])
p.poligons()
plt.grid(True)
plt.show()
```
